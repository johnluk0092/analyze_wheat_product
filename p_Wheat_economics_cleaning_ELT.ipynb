# function transform data

```python
def transform_long_to_wide(df_long, year_col='year', region_col='region', 
                          column_col='column', value_col='value', 
                          fill_missing_years=True, fill_value=np.nan):
    
    # Get unique values
    actual_years = set(df_long[year_col].unique())
    all_regions = df_long[region_col].unique()
    all_columns = df_long[column_col].unique()
    
    # Determine year range
    min_year = min(actual_years)
    max_year = max(actual_years)
    
    if fill_missing_years:
        all_years = range(min_year, max_year + 1)
    else:
        all_years = sorted(actual_years)
    
    # Create an empty dataframe for results
    result_data = []
    
    for year in all_years:
        for region in all_regions:
            # Create a row dictionary
            row = {year_col: year, region_col: region}
            
            # Check if this year exists in data
            if year in actual_years:
                # Year exists - try to get data for each column
                for col in all_columns:
                    filtered = df_long[(df_long[column_col] == col) & 
                                      (df_long[year_col] == year) & 
                                      (df_long[region_col] == region)]
                    
                    if not filtered.empty:
                        row[col] = filtered[value_col].iloc[0]
                    else:
                        row[col] = fill_value
            else:
                # Year doesn't exist in data - fill all with fill_value
                for col in all_columns:
                    row[col] = fill_value
            
            result_data.append(row)
    
    # Convert to DataFrame
    result_df = pd.DataFrame(result_data)
    
    # Reorder columns: year, region, then alphabetically sorted other columns
    other_cols = sorted([col for col in result_df.columns if col not in [year_col, region_col]])
    result_df = result_df[[year_col, region_col] + other_cols]
    
    return result_df
```

```python
df_a_1 = transform_long_to_wide(df_a, year_col='year', region_col='region', 
                          column_col='column', value_col='value', 
                          fill_missing_years=True, fill_value=np.nan)
file_path = ".\data_warehouse\wheat_dollars_per_planted_acre.gzip"
df_a_1.to_parquet(file_path)
print(f"\nDataFrame successfully saved to {file_path}")
```
